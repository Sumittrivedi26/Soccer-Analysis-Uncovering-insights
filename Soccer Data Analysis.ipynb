```python
# creating a new column

data['MinPerMatch'] = (data['Mins']/data['Matches']).astype(int)

data.head()
```

```python
#Total Goals in that season

total_goals = data['Goals'].sum()

print(total_goals)
```

```python
# total Penalty Goals in the season

total_penalty_goals = data['Penalty_Goals'].sum()

print(total_penalty_goals)
```

```python
# total penalty goals

total_penalty_attempted = data['Penalty_Attempted'].sum()

print(total_penalty_attempted)
```

##### From the above analysis we can say that a total of 125 penalty attempts were taken throughout the season out of which 102 were converted into penalt goals rest of which were missed 

```python
plt.figure(figsize = (5,5))
plt_not_scored = total_penalty_attempted - total_penalty_goals
plt_not_scored
pie_data =[plt_not_scored, total_penalty_goals]
labels = ['Penalties_missed', 'Penalty_goals']
plt.pie(pie_data, labels = labels, autopct = '%0.0f%%')
plt.show
```

```python
### age bins

under20 = data[data['Age'] < 20]
between20_25 = data[(data['Age'] >= 20) & (data['Age'] <= 25)]
between25_30 = data[(data['Age'] >= 25) & (data['Age'] <= 30)]
above30 = data[data['Age'] > 30]
```

```python
x = [under20['Name'].count(), between20_25['Name'].count(), between25_30['Name'].count(), above30['Name'].count()]
label = ['under20', 'between20_25', 'between25_30', 'above30']
df = pd.DataFrame(x, index = label )


ax= df.plot(kind = 'barh', color = sns.color_palette('Set2'), figsize = (8,5))

ax.bar_label(ax.containers[0])
plt.show
```
